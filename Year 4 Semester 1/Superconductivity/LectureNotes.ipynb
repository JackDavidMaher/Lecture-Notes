## Thermodynamics of Superconductors
### Free Energy
$$S: B=0$$
$$n: \chi=0$$
For a superconductor,
$G_S(T,H) - G_S(T,0) = \int_0^H dG = - \int_0^H \mu_0 M dH = \int_0^H \mu_0 H dH = \frac{1}{2} \mu_0 H^2 $

For a metal,
$G_n(T,H) - G_n(T,0) = -\int_0^H \mu_0MdH = -\frac{1}{2} \chi_P \mu_0 H^2 =0 $

$$ dG = -SdT - \mu_0 M dH $$
At fixed temperature,
$$ \left(\frac{\partial G}{\partial H}\right)_T = -\mu_0 M $$
### Entropy
Phase boundary $H_C(T)$ is defined by the condition
$$ G_S(T,H_C) = G_n(T,H_C) $$
$$ G_S(T,0) - G_n(T,0) = -\frac{1}{2} \mu_0 H_C^2 $$
As one lowers T, more energy gain to become SC, therfore H_C increases with lowering T.

$$ S_S(T,H_\alpha) = -\frac{dG_S(T,H_\alpha)}{dT} $$
$$ = -\frac{d}{dT}[G_\eta(T,H_\alpha)-\frac{\mu_0}{2}(H_C^2-H_\alpha^2)] $$
$$ = -\frac{dG_\eta(T,0)}{dT} + \frac{d}{dT} \left(\frac{\mu_0}{2}(H_C^2-H_\alpha^2)\right) $$
$$ = S_\eta(T,0) + \mu_0 H_C \frac{dH_C}{dT} $$

$$

## Recap
Cooper pair problem: an electron interacting with another electron throguh a potential $-|V|$ mediated by phonons with energy less than the Debye frequency $\hbar \omega_D$

the reduced interaction is isotropic in momentum space

we assume the electrons have properties of kinetic energies $k$ and $-k$, forming a Cooper pair with total momentum zero and if one electron is spin up, the other is spin down.

$$ H = \sum_{k,\sigma} \xi_k C_{k,\sigma}^\dagger C_{k,\sigma} - \sum_{k,k'} |V|_{kk'} C_{k\uparrow}^\dagger C_{-k\downarrow}^\dagger C_{-k'\downarrow} C_{k'\uparrow} $$
This is the BCS Hamiltonian describing electrons with kinetic energies $\xi_k$ interacting via an attractive potential $-|V|$ that leads to the formation of Cooper pairs. 

The Hamiltonian consists of two parts: the first term represents the kinetic energy of electrons relative to the chemical potential, and the second term represents the attractive interaction between electrons that can form Cooper pairs.

The Hamiltonian is central to the BCS theory of superconductivity, as it captures the essential physics of electron pairing and the resulting superconducting state.

$$|\text{BCS}\rangle = \prod_k (u_k + v_k C_{k\uparrow}^\dagger C_{-k\downarrow}^\dagger) |\text{vac}\rangle $$
where

$ u_k $ and $ v_k $ are the coherence factors,
$ |\text{vac}\rangle $ is the vacuum state,
$ k $ is the momentum of the particles.
$ u_k $ and $ v_k $ satisfy the normalization condition $ |u_k|^2 + |v_k|^2 = 1 $.

This equation represents the BCS ground state, which is a coherent superposition of states with different numbers of Cooper pairs. It captures the essential feature of superconductivity, where electrons form pairs and condense into a macroscopic quantum state.

By adding electrons, one by one, into the system, the BCS ground state allows for the formation of Cooper pairs at each momentum state $k$. The coherence factors $u_k$ and $v_k$ determine the probability amplitudes for the states being empty or occupied by a Cooper pair, respectively. This construction ensures that the ground state is a superposition of all possible configurations of Cooper pairs, leading to the macroscopic quantum coherence characteristic of superconductivity.

$$ |\text{F.S}\rangle = \prod_{|k|<k_F} C_{k\uparrow}^\dagger C_{k\downarrow}^\dagger |\text{vac}\rangle $$

We are dealing with the grand canonical ensemble, where the number of particles is not fixed, and the system can exchange particles with a reservoir. In this context, the Fermi sea state $|\text{F.S}\rangle$ represents the filled states up to the Fermi level $k_F$, and it serves as the reference state for constructing the BCS ground state.

$$ n_k = C_{k\uparrow}^\dagger C_{k\uparrow}$$
$$ \bar{N} = \sum_k 2 |v_k|^2 $$

The average number of particles $\bar{N}$ in the BCS ground state is given by the sum over all momentum states of twice the probability of each state being occupied by a Cooper pair, which is $2|v_k|^2$. This reflects the fact that each Cooper pair consists of two electrons.

$$ H_\text{BCS}^\text{MF} = \sum_{k\sigma} \xi_k (C_{k\sigma}^\dagger C_{k\sigma}) - \Delta \sum_k (C_{k\uparrow}^\dagger C_{-k\downarrow}^\dagger + C_{-k\downarrow} C_{k\uparrow}) + \frac{\Delta^2}{V}$$
Where

$$ \xi_k = \epsilon_k - \mu $$
is the single-particle energy relative to the chemical potential, and
$$ \Delta = |V| \sum_{k'} \langle C_{-k'\downarrow} C_{k'\uparrow} \rangle $$
is the superconducting gap parameter, which characterizes the strength of the Cooper pairing in the system. The mean-field Hamiltonian $H_\text{BCS}^\text{MF}$ describes the effective behavior of electrons in the superconducting state, taking into account the formation of Cooper pairs.
