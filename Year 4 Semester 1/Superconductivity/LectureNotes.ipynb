We have probes to understand the nature of superconductive materials, such as:
- Transport: $\sigma(1/\rho)$, k, ...
- Thermodynamic: $C_p$, $M$, $S$, ...
- Scattering: X-ray (Structure information of whole material)
- Imaging: Electron microscopy (Structure at atomic scale)

$T_c$ 1 ~ 10 K - Critical temperature (Superconducting transition)

$T_F$ ~ 10000 K - Fermi temperature (Electron contribution)

$\Theta_D$ ~ 100 K - Debye temperature (Phonon contribution)

Flux quantum: $\Phi_0 = \frac{h}{2e}$ - Quantum of magnetic flux in a superconductor

Current can occur spontaneously in a superconductor without any applied voltage due to the formation of Cooper pairs. "WTF"

Alternating current (AC) can also flow without resistance in a superconductor with voltage applied.
