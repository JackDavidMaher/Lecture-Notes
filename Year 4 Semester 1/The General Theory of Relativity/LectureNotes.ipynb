## Path Length and Lagrangian
### Metric along the path $x^\mu = x^\mu(\tau)$
$$ ds^2 = g_{\mu\nu} dx^\mu dx^\nu = g_{\mu\nu} \frac{dx^\mu}{d\tau} d\tau \frac{dx^\nu}{d\tau} d\tau = g_{\mu\nu}\dot{x}^\mu \dot{x}^\nu d\tau^2 $$
$$ \frac{ds}{d\tau} = \sqrt{g_{\mu\nu}\dot{x}^\mu \dot{x}^\nu} = \hat{\mathcal{L}} $$
### Action
$$ S = \int ds = \int \hat{\mathcal{L}} d\tau = \int \sqrt{g_{\mu\nu}\dot{x}^\mu \dot{x}^\nu} d\tau $$
