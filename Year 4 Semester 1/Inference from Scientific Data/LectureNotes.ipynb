### Bayes' Theorem
- M - Model
- D - Data
- $\theta$ - Parameters of the model
$$P(M|D) = \frac{P(D|M)P(M)}{P(D)}$$
- $P(M)$ is the prior probability of the model.
- $P(D|M)$ is the likelihood of the data given the model.
- $P(D)$ is the evidence.
- $P(M|D)$ is the posterior probability of the model given the data.
Model space represents our prior. It is drawn as a square with shapes inside representing different models.

We move from model space into data space with liklihood ($P(D|M_i)$) for each model $M_i$, and then use Bayes' theorem to update our beliefs about the models based on the observed data.

$$ P(M|D) \propto P(D|M)P(M) $$

#### Example
4 models:
- $P(D|M_0)$ is small
- $P(D|M_1)$ is large
- $P(D|M_2)$ is small
- $P(D|M_3)$ is medium
Given the likelihoods, we can update our beliefs about the models using Bayes' theorem. Models with higher likelihoods will have thier posterior probabilities increased.

This same process can be applied to parameter estimation (those contour plots), where we update our beliefs about the parameters $\theta$ of a model based on observed data our prior then becomes a new posterior (another contour plot).

$$ P(\theta|D) \propto P(D|\theta)P(\theta) $$

#### Example
$\theta$ - $S,\bar{S}$

$D$ - $W,\bar{W}$

$P(S)=0.2$,$P(\bar{S})=0.8$,$P(W|S)=0.5$,$P(W|\bar{S})=0.01$

Now this looks good but we dont have the evidence needed to apply Bayes' theorem. We need to calculate the evidence $P(D)$ first:
$$ P(D) = \int P(D|\theta)P(\theta) d\theta $$
For our discrete example:
$$ P(W) = \sum_{\theta} P(W|\theta)P(\theta) = P(W|S)P(S) + P(W|\bar{S})P(\bar{S}) = 0.5 \cdot 0.2 + 0.01 \cdot 0.8 = 0.1 + 0.008 = 0.108 $$
Now we can apply Bayes' theorem to calculate the posterior probability of $\theta$ given the data $D=W$:
$$ P(S|W) = \frac{P(W|S)P(S)}{P(W)} = \frac{0.5 \cdot 0.2}{0.108} \approx 0.926 $$
$$ P(\bar{S}|W) = \frac{P(W|\bar{S})P(\bar{S})}{P(W)} = \frac{0.01 \cdot 0.8}{0.108} \approx 0.074 $$

## Bayesian Inference
Posterior = Likelihood * Prior / Evidence

What I believe after seeing the data = How well the model predicts the data X What I believed before seeing the data / How predictable the data was before seeing it

### Parameter Estimation
$$ \theta = (\phi, \psi) $$
$\phi$ is the parameter of interest, and $\psi$ represents nuisance parameters.
$$ p(\phi|D) \propto \int p(D|\phi, \psi) p(\phi, \psi) d\psi $$

$$ p(D) = \int p(D|\theta) p(\theta) d\theta $$
If $\theta \subset \mathbb{R}^k$, then the integral is over the entire parameter space.
$$ p(D) = \int_{\mathbb{R}^k} p(D|\theta_1,...,\theta_k) p(\theta_1,...,\theta_k) d\theta_1 ... d\theta_k $$
### Bayesian Decision Making Workflow
    - Prior
    - Likelihood
    - Posterior
    - Utility
    - Decision
## Lego Example
### Prior
We are assuming a prior belief about the child's age, which could be represented as a probability distribution over possible ages.
For this example I decided to use birth rate data to calculate the prior probability distribution over the child's age.

### Likelihood
The likelihood represents how probable the observed data is given different possible ages of the child. In this example, it could be based on a range of observed behaviors or characteristics that are age-dependent.

### Posterior
The posterior combines the prior and the likelihood to give an updated belief about the child's age after observing the data. It is calculated using Bayes' theorem:
$$ p(\text{age} | \text{data}) \propto p(\text{data} | \text{age}) p(\text{age}) $$

### Utility
Actions $A={a_1, a_2, ..., a_n}$

where each action $a_i$ corresponds to a possible decision we can make regarding the child's age.

Utility $u(a,\theta)$ represents the value or payoff associated with taking action $a$ when the true state of the world is $\theta$.

The utility represents the value or payoff associated with different decisions. In this example, it could be related to the accuracy of predicting the child's age or the consequences of making an incorrect prediction.
$$ \mathbb{E}_{\theta|D}[u(a|D)] = \int u(a|\theta) p(\theta|D) d\theta $$

### Decision
The decision is made by combining the posterior and the utility to choose the action that maximizes expected utility. For instance, we might choose the age that has the highest posterior probability or that maximizes expected utility given the potential outcomes.
