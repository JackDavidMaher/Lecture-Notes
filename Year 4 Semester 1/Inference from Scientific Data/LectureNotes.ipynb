### Bayes' Theorem
- M - Model
- D - Data
- $\theta$ - Parameters of the model
$$P(M|D) = \frac{P(D|M)P(M)}{P(D)}$$
- $P(M)$ is the prior probability of the model.
- $P(D|M)$ is the likelihood of the data given the model.
- $P(D)$ is the evidence.
- $P(M|D)$ is the posterior probability of the model given the data.
Model space represents our prior. It is drawn as a square with shapes inside representing different models.

We move from model space into data space with liklihood ($P(D|M_i)$) for each model $M_i$, and then use Bayes' theorem to update our beliefs about the models based on the observed data.

$$ P(M|D) \propto P(D|M)P(M) $$

#### Example
4 models:
- $P(D|M_0)$ is small
- $P(D|M_1)$ is large
- $P(D|M_2)$ is small
- $P(D|M_3)$ is medium
Given the likelihoods, we can update our beliefs about the models using Bayes' theorem. Models with higher likelihoods will have thier posterior probabilities increased.

This same process can be applied to parameter estimation (those contour plots), where we update our beliefs about the parameters $\theta$ of a model based on observed data our prior then becomes a new posterior (another contour plot).

$$ P(\theta|D) \propto P(D|\theta)P(\theta) $$

#### Example
$\theta$ - $S,\bar{S}$

$D$ - $W,\bar{W}$

$P(S)=0.2$,$P(\bar{S})=0.8$,$P(W|S)=0.5$,$P(W|\bar{S})=0.01$

Now this looks good but we dont have the evidence needed to apply Bayes' theorem. We need to calculate the evidence $P(D)$ first:
$$ P(D) = \int P(D|\theta)P(\theta) d\theta $$
For our discrete example:
$$ P(W) = \sum_{\theta} P(W|\theta)P(\theta) = P(W|S)P(S) + P(W|\bar{S})P(\bar{S}) = 0.5 \cdot 0.2 + 0.01 \cdot 0.8 = 0.1 + 0.008 = 0.108 $$
Now we can apply Bayes' theorem to calculate the posterior probability of $\theta$ given the data $D=W$:
$$ P(S|W) = \frac{P(W|S)P(S)}{P(W)} = \frac{0.5 \cdot 0.2}{0.108} \approx 0.926 $$
$$ P(\bar{S}|W) = \frac{P(W|\bar{S})P(\bar{S})}{P(W)} = \frac{0.01 \cdot 0.8}{0.108} \approx 0.074 $$
