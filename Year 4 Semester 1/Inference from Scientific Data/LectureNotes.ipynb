### Bayes' Theorem
$$\sum_i P(A_i) = 1$$
$$P(B) = \sum_i P(B|A_i)P(A_i)$$

$$P(A,B) = P(B)P(A|B)$$
$$P(B,A) = P(A)P(B|A)$$
$$P(A,B) = P(B,A)$$
$$P(A|B)P(B) = P(B|A)P(A)$$
$$P(A|B) = \frac{P(B|A)P(A)}{P(B)}$$

$$P(\theta|M,D) = \frac{P(M,D|\theta)P(\theta)}{P(D,M)}$$
$$P(\theta|D) = \frac{P(D|\theta)P(\theta)}{P(D)}$$
$$P(M|D) = \frac{P(D|M)P(M)}{P(D)}$$
