#### Bayes' Theorem
$$ P(\lambda|D) = \frac{P(D|\lambda) P(\lambda)}{P(D)} $$
#### Likelihood
$$ P(D|\lambda) = \prod_i^n P(x_i|\lambda) = \prod_i^n \lambda e^{-\lambda x_i} = \lambda^n e^{-\lambda \sum_i^n x_i} $$
$$ S = \sum_i^n x_i $$
$$ P(D|\lambda) = \lambda^n e^{-\lambda S}$$
#### Prior
$$ P(\lambda) = \frac{\beta^\alpha}{\Gamma(\alpha)} \lambda^{\alpha-1} e^{-\beta \lambda} $$
#### Posterior
$$ P(\lambda|D) = \frac{P(D|\lambda) P(\lambda)}{P(D)} = \frac{(\lambda^n e^{-\lambda S})( \frac{\beta^\alpha}{\Gamma(\alpha)} \lambda^{\alpha-1} e^{-\beta \lambda})}{P(D)} = \frac{\lambda^{n+\alpha-1} e^{-\lambda (S+\beta)}}{\Gamma(\alpha) P(D)} $$
#### Normalisation
$$ \int_{-\infty}^{\infty} P(\lambda|D) \, d\lambda = 1 $$
$$ \int_{-\infty}^{\infty} \frac{\lambda^{n+\alpha-1} e^{-\lambda (S+\beta)}}{\Gamma(\alpha) P(D)} \, d\lambda = 0 + \int_{0}^{\infty} \frac{\lambda^{n+\alpha-1} e^{-\lambda (S+\beta)}}{\Gamma(\alpha) P(D)} \, d\lambda = 1 $$
$$ \int_{0}^{\infty}x^{a} e^{bx} \, dx = \frac{\Gamma(a+1)}{(-b)^{a+1}} $$
$$ a = n + \alpha - 1, b = -(S+\beta) $$
$$ \int_{0}^{\infty} \lambda^{n+\alpha-1} e^{-\lambda (S+\beta)} \, d\lambda = \frac{\Gamma(n+\alpha)}{(S+\beta)^{n+\alpha}} $$
$$ P(D) = \frac{\Gamma(n+\alpha)}{\Gamma(\alpha) (S+\beta)^{n+\alpha}} $$
$$ P(\lambda|D) = \frac{(S+\beta)^{n+\alpha}}{\Gamma(n+\alpha)} \lambda^{n+\alpha-1} e^{-\lambda (S+\beta)} $$

### \[2]
$$ \mathbb{E}[\lambda|D] = \int_{0}^{\infty} \lambda P(\lambda|D) \, d\lambda $$
$$ \mathbb{E}[\lambda|D] = \int_{0}^{\infty} \lambda \frac{(S+\beta)^{n+\alpha}}{\Gamma(n+\alpha)} \lambda^{n+\alpha-1} e^{-\lambda (S+\beta)} \, d\lambda $$
$$ \mathbb{E}[\lambda|D] = \frac{(S+\beta)^{n+\alpha}}{\Gamma(n+\alpha)} \int_{0}^{\infty} \lambda^{n+\alpha} e^{-\lambda (S+\beta)} \, d\lambda $$
$$ \mathbb{E}[\lambda|D] = \frac{(S+\beta)^{n+\alpha}}{\Gamma(n+\alpha)} \frac{\Gamma(n+\alpha+1)}{(S+\beta)^{n+\alpha+1}} $$
$$ \mathbb{E}[\lambda|D] = \frac{\Gamma(n+\alpha+1)}{\Gamma(n+\alpha) (S+\beta)} $$
$$ \mathbb{E}[\lambda|D] = \frac{n+\alpha}{S+\beta} $$

```python
import numpy as np
import matplotlib.pyplot as plt
from math import gamma
x = np.linspace(-5, 15, 100)
def gamma_pdf(x1, k1, theta):
    return np.where(x1>0, ((x1 ** (k1 - 1)) * np.exp(-x1 / theta)) / (gamma(k1) * theta ** k1), 0)
plt.plot(x, gamma_pdf(x, 1, 1), label=r'$(k_1,\theta)=(1,1)$')
plt.plot(x, gamma_pdf(x, 2, 1), label=r'$(k_1,\theta)=(2,1)$')
plt.plot(x, gamma_pdf(x, 2, 2), label=r'$(k_1,\theta)=(2,2)$')

plt.title('Question 2.1  -  Gamma PDFs')
plt.legend()
plt.show()
```
