#### Bayes Theorem
$$ P(\lambda|D) = \frac{P(D|\lambda) P(\lambda)}{P(D)} $$
#### Using given information to simplify the expression for the likelihood
$$ P(D|\lambda) = \prod_i^n P(x_i|\lambda) = \prod_i^n \lambda e^{-\lambda x_i} = \lambda^n e^{-\lambda \sum_i x_i} $$
$$ S = \sum_i^n x_i $$
$$ P(\lambda|D) = \lambda^n e^{-\lambda S}$$
#### Prior
$$ P(\lambda) = \frac{\beta^\alpha}{\Gamma(\alpha)} \lambda^{\alpha-1} e^{-\beta \lambda} $$
#### Posterior
$$ P(\lambda|D) = \frac{P(D|\lambda) P(\lambda)}{P(D)} = \frac{(\lambda^n e^{-\lambda S})( \frac{\beta^\alpha}{\Gamma(\alpha)} \lambda^{\alpha-1} e^{-\beta \lambda})}{P(D)} = \frac{\lambda^{n+\alpha-1} e^{-\lambda (S+\beta)}}{\Gamma(\alpha) P(D)} $$

```python
import numpy as np
import matplotlib.pyplot as plt
from math import gamma
x = np.linspace(-5, 10, 1001)
def gamma_pdf(x1, k, theta):
    return np.where(x1>0, ((x1 ** (k - 1)) * np.exp(-x1 / theta)) / (gamma(k) * theta ** k), 0)
plt.plot(x, gamma_pdf(x, 1, 1), label=r'$(k_1,\theta)=(1,1)$')
plt.plot(x, gamma_pdf(x, 2, 1), label=r'$(k_1,\theta)=(2,1)$')
plt.plot(x, gamma_pdf(x, 2, 2), label=r'$(k_1,\theta)=(2,2)$')

plt.title('Question 2.1  -  Gamma PDFs')
plt.legend()
plt.grid()
plt.show()
```
