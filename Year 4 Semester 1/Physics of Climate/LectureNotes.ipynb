### Equilibrium Temperature
Climate/weather is moving energy around the planet
- What is our energy budget?
Assume here that all incoming energy is from host star

Star has radius $(R_*)$ and temperature $(T_\text{eff})$ and planet has radius $(R_p)$ and temperature $(T_p)$ at distance $(a)$
$$ L = A \sigma T^4 $$

Star at surface: $L_* = 4 \pi R_*^2 \sigma T_\text{eff}^4$

Reaching the planet: $L_\text{in} = \frac{L_*}{4 \pi a^2} \pi R_p^2$

Planet emits: $L_\text{out} = 4 \pi R_p^2 \sigma T_p^4$

At equilibrium: $L_\text{in} = L_\text{out}$

Eq 1.1:
$$ T_p = \sqrt{\frac{R_*}{2a}} T_\text{eff} = T_\text{eq} = \text{equilibrium temperature} $$

#### Bond Albedo
The Bond albedo $(A_B)$ is the fraction of incident energy that is scattered/reflected accross all wavelengths and directions.
The equilibrium temperature considering the Bond albedo is:

Earth; $A_B \approx 0.3$
$$ L_\text{in} = \frac{L_*}{4 \pi a^2} \pi R_p^2 (1-A_B) $$
$$ T_\text{eq} = (1-A_B)^{1/4} \sqrt{\frac{R_*}{2a}} T_\text{eff} $$

##### Earth
- Bond albedo: $A_B \approx 0.3$
- Equilibrium temperature: $T_\text{eq} \approx 255 \text{ K}$
##### Venus
- Bond albedo: $A_B \approx 0.75$
- Equilibrium temperature: $T_\text{eq} \approx 232 \text{ K}$

### Simple Climate Model
A slab of atmosphere with area $(A)$ and temperature $(T_a)$
- Transparent in optical, opaque in infrared
- Sunlight goes straight through to the surface
- Surface re-emits infrared radiation
- Atmosphere is NOT a blackbody radiator
- Modify luminosity by $ \epsilon_\text{IR} \leq 1 $, the emissivity of the atmosphere
    - $\sigma T_a^4 \rightarrow \epsilon_\text{IR} \sigma T_a^4$
    - $L_\text{out} = 2A \epsilon_\text{IR} \sigma T_a^4$
    - $L_\text{in} = A \epsilon_\text{IR} \sigma T_s^4$

Eq 1.3:
$$ T_a = 2^{-1/4} T_s $$

Surface energy balance:
$$ \sigma T_\text{eq}^4 + \epsilon_\text{IR} \sigma T_a^4 = \sigma T_s^4 $$

Eq 1.5:
$$ T_s = \frac{T_\text{eq}}{(1 - \epsilon_\text{IR}/2)^{1/4}} $$

If $\epsilon_\text{IR} = 1$, the atmosphere is a perfect blackbody in the infrared:
$$ T_s = 2^{1/4} T_\text{eq} $$

If $\epsilon_\text{IR} = 0$, the atmosphere does not emit in the infrared:
$$ T_s = T_\text{eq} $$

##### With Opaque Atmosphere
- Earth surface temperature: $T_s \approx 303 \text{ K}$
- Venus surface temperature: $T_s \approx 276 \text{ K}$
