### Equilibrium Temperature
Climate/weather is moving energy around the planet
- What is our energy budget?
Assume here that all incoming energy is from host star

Star has radius $(R_*)$ and temperature $(T_\text{eff})$ and planet has radius $(R_p)$ and temperature $(T_p)$ at distance $(a)$
$$ L = A \sigma T^4 $$

Star at surface: $L_* = 4 \pi R_*^2 \sigma T_\text{eff}^4$

Reaching the planet: $L_\text{in} = \frac{L_*}{4 \pi a^2} \pi R_p^2$

Planet emits: $L_\text{out} = 4 \pi R_p^2 \sigma T_p^4$

At equilibrium: $L_\text{in} = L_\text{out}$

Eq 1.1:
$$ T_p = \sqrt{\frac{R_*}{2a}} T_\text{eff} = T_\text{eq} = \text{equilibrium temperature} $$

#### Bond Albedo
The Bond albedo $(A_B)$ is the fraction of incident energy that is scattered/reflected accross all wavelengths and directions.
The equilibrium temperature considering the Bond albedo is:

Earth; $A_B \approx 0.3$
$$ L_\text{in} = \frac{L_*}{4 \pi a^2} \pi R_p^2 (1-A_B) $$
$$ T_\text{eq} = (1-A_B)^{1/4} \sqrt{\frac{R_*}{2a}} T_\text{eff} $$

##### Earth
- Bond albedo: $A_B \approx 0.3$
- Equilibrium temperature: $T_\text{eq} \approx 255 \text{ K}$
##### Venus
- Bond albedo: $A_B \approx 0.75$
- Equilibrium temperature: $T_\text{eq} \approx 232 \text{ K}$

### Simple Climate Model
A slab of atmosphere with area $(A)$ and temperature $(T_a)$
- Transparent in optical, opaque in infrared
- Sunlight goes straight through to the surface
- Surface re-emits infrared radiation
- Atmosphere is NOT a blackbody radiator
- Modify luminosity by $ \epsilon_\text{IR} \leq 1 $, the emissivity of the atmosphere
    - $\sigma T_a^4 \rightarrow \epsilon_\text{IR} \sigma T_a^4$
    - $L_\text{out} = 2A \epsilon_\text{IR} \sigma T_a^4$
    - $L_\text{in} = A \epsilon_\text{IR} \sigma T_s^4$

Eq 1.3:
$$ T_a = 2^{-1/4} T_s $$

Surface energy balance:
$$ \sigma T_\text{eq}^4 + \epsilon_\text{IR} \sigma T_a^4 = \sigma T_s^4 $$

Eq 1.5:
$$ T_s = \frac{T_\text{eq}}{(1 - \epsilon_\text{IR}/2)^{1/4}} $$

If $\epsilon_\text{IR} = 1$, the atmosphere is a perfect blackbody in the infrared:
$$ T_s = 2^{1/4} T_\text{eq} $$

If $\epsilon_\text{IR} = 0$, the atmosphere does not emit in the infrared:
$$ T_s = T_\text{eq} $$

##### With Opaque Atmosphere
- Earth surface temperature: $T_s \approx 303 \text{ K}$
- Venus surface temperature: $T_s \approx 276 \text{ K}$

### Hydrostatic equilibrium
$$ A(P+ \Delta P) + \rho g A \Delta z = A P$$
$$ \frac{dP}{dz} = -\rho g $$
$$ P=\frac{k_B T \rho}{\mu} $$
$$ \frac{dP}{dz} = -\frac{\mu g}{k_B T} P , \int_{P_0}^{P} d \ln P' = \int_{0}^{z} -\frac{\mu g}{k_B T} dz' $$
$$ \ln \frac{P}{P_0} = -\frac{\mu g}{k_B T} z $$
$$ P = P_0 \exp \left( -\frac{\mu g}{k_B T} z \right) $$
$$ H_{sc} = \frac{k_B T}{\mu g} $$
$$ P = P_0 \exp \left( -\frac{z}{H_{sc}} \right) $$
### Ideal gas law
$$ P = nk_bT $$
$$ P = \frac{k_B}{\mu} \rho T $$
$$ PV = Nk_B T $$
$$ \ln P = \ln n + \ln T + const. $$
$$ d\ln P = d\ln n + d\ln T $$
#### Useful relations

$C_P,C_V$ - heat capacities at constant pressure and volume

$c_p,c_v$ - specific heat capacities at constant pressure and volume

$\gamma = \frac{c_p}{c_v}$ - adiabatic index

$N_{mol}R = C_P - C_V$ - Mayer's relation

$c_P - c_v = \frac{N_{mol} R}{M} = \frac{k_B}{\mu}$

##### Adiabatic gas
$$ PV^\gamma = \text{constant} $$
$$ T V^{\gamma-1} = \text{constant} $$

## Recap
- Hydrostatic equilibrium $\frac{dP}{dz} = -\rho g$
- Atmoshpheric scale height $H = \frac{k_B T}{\mu g}$
- Convective stability $\frac{dlnT}{dlnP} < \frac{\gamma -1}{\gamma}$
## Dry convection
When convection happens, $\frac{dlnT}{dlnP} = \frac{\gamma -1}{\gamma}$
### Lapse rate
Lapse rate = $-\frac{dT}{dz}$
$$ \frac{dP}{dz} = -\rho g $$
$$ P = \frac{\rho}{\mu} k_B T $$
$$ \frac{dP}{dz} = - \frac{\mu g}{k_B T} P $$
$$ T \frac{dP}{P} = - \frac{\mu g}{k_B} dz $$
$$ PV^\gamma = \text{constant} $$
$$ P^{1-\gamma} T^\gamma = \text{constant} $$
$$ \gamma ln(T) = (\gamma -1) ln(P) + \text{constant} \to \gamma \frac{dT}{T} = (\gamma -1) \frac{dP}{P} $$
$$ -\frac{dT}{dz} = \frac{\gamma -1}{\gamma} \frac{\mu g}{k_B} $$
$$ \frac{\gamma -1}{\gamma} = \frac{\frac{C_P}{C_V}-1}{\frac{C_P}{C_V}} = \frac{C_P - C_V}{C_P} \to \frac{k_B/\mu}{C_P} $$
$$ -\frac{dT}{dz} = \frac{k_B/\mu}{C_P} \frac{\mu g}{k_B} = \frac{g}{C_P} $$
## Partial Pressure
$P_i$ is the partial pressure of component $i$ in a mixture of gases.
$ P_{O_2} = 0.21 atm \text{ (on earth)}$
$$ \frac{P_i}{P} = \frac{n_i}{n} = \frac{n_ik_BT}{n k_B T}$$
$$ \sum_i P_i = P $$
### Saturation Vapor Pressure
$$ \frac{dP_{sat}}{dT} = \frac{1}{T} \frac{L}{\rho_v^{-1}-\rho_c^{-1}} $$
$$ \rho_c \gg \rho_v $$
$$ \frac{dP_{sat}}{dT} = \frac{L\mu}{k_BT^2}P_{sat} $$
## Single-Component Condensable Atmosphere
$$ P_{sat} = P_{sat}(T) e^{-\frac{L\mu}{k_B} \left( \frac{1}{T}  - \frac{1}{T_0}\right)} $$
$$ T(P) = \frac{T_0}{1 - \frac{k_BT_0}{L\mu} \ln \frac{P}{P_{sat}(T_0)}} $$

Dry adiabat: gradient of temperature with height in a dry atmosphere. Not absolute temperature

Wet adiabat: gradient and pressure-temperature location is fixed (fundamental property of material)
