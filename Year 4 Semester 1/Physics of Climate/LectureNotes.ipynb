### Equilibrium Temperature
Climate/weather is moving energy around the planet
- What is our energy budget?
Assume here that all incoming energy is from host star

Star has radius $(R_*)$ and temperature $(T_\text{eff})$ and planet has radius $(R_p)$ and temperature $(T_p)$ at distance $(a)$
$$ L = A \sigma T^4 $$

Star at surface: $L_* = 4 \pi R_*^2 \sigma T_\text{eff}^4$

Reaching the planet: $L_\text{in} = \frac{L_*}{4 \pi a^2} \pi R_p^2$

Planet emits: $L_\text{out} = 4 \pi R_p^2 \sigma T_p^4$

At equilibrium: $L_\text{in} = L_\text{out}$

Eq 1.1:
$$ T_p = \sqrt{\frac{R_*}{2a}} T_\text{eff} = T_\text{eq} = \text{equilibrium temperature} $$

#### Bond Albedo
The Bond albedo $(A_B)$ is the fraction of incident energy that is scattered/reflected accross all wavelengths and directions.
The equilibrium temperature considering the Bond albedo is:

Earth; $A_B \approx 0.3$
$$ L_\text{in} = \frac{L_*}{4 \pi a^2} \pi R_p^2 (1-A_B) $$
$$ T_\text{eq} = (1-A_B)^{1/4} \sqrt{\frac{R_*}{2a}} T_\text{eff} $$

##### Earth
- Bond albedo: $A_B \approx 0.3$
- Equilibrium temperature: $T_\text{eq} \approx 255 \text{ K}$
##### Venus
- Bond albedo: $A_B \approx 0.75$
- Equilibrium temperature: $T_\text{eq} \approx 232 \text{ K}$

### Simple Climate Model
A slab of atmosphere with area $(A)$ and temperature $(T_a)$
- Transparent in optical, opaque in infrared
- Sunlight goes straight through to the surface
- Surface re-emits infrared radiation
- Atmosphere is NOT a blackbody radiator
- Modify luminosity by $ \epsilon_\text{IR} \leq 1 $, the emissivity of the atmosphere
    - $\sigma T_a^4 \rightarrow \epsilon_\text{IR} \sigma T_a^4$
    - $L_\text{out} = 2A \epsilon_\text{IR} \sigma T_a^4$
    - $L_\text{in} = A \epsilon_\text{IR} \sigma T_s^4$

Eq 1.3:
$$ T_a = 2^{-1/4} T_s $$

Surface energy balance:
$$ \sigma T_\text{eq}^4 + \epsilon_\text{IR} \sigma T_a^4 = \sigma T_s^4 $$

Eq 1.5:
$$ T_s = \frac{T_\text{eq}}{(1 - \epsilon_\text{IR}/2)^{1/4}} $$

If $\epsilon_\text{IR} = 1$, the atmosphere is a perfect blackbody in the infrared:
$$ T_s = 2^{1/4} T_\text{eq} $$

If $\epsilon_\text{IR} = 0$, the atmosphere does not emit in the infrared:
$$ T_s = T_\text{eq} $$

##### With Opaque Atmosphere
- Earth surface temperature: $T_s \approx 303 \text{ K}$
- Venus surface temperature: $T_s \approx 276 \text{ K}$

### Hydrostatic equilibrium
$$ A(P+ \Delta P) + \rho g A \Delta z = A P$$
$$ \frac{dP}{dz} = -\rho g $$
$$ P=\frac{k_B T \rho}{\mu} $$
$$ \frac{dP}{dz} = -\frac{\mu g}{k_B T} P , \int_{P_0}^{P} d \ln P' = \int_{0}^{z} -\frac{\mu g}{k_B T} dz' $$
$$ \ln \frac{P}{P_0} = -\frac{\mu g}{k_B T} z $$
$$ P = P_0 \exp \left( -\frac{\mu g}{k_B T} z \right) $$
$$ H_{sc} = \frac{k_B T}{\mu g} $$
$$ P = P_0 \exp \left( -\frac{z}{H_{sc}} \right) $$

## Optical Depth
Optical depth is the fractional area blocked by particles: $\frac{N\sigma}{A} = \frac{n\sigma A \delta s}{A} = n\sigma \delta s$

$d \tau = \sigma n ds = \kappa \rho ds$, $\kappa$ is the mass absorption coefficient ($\frac{\sigma}{\mu}$)

$d\tau$ can be integrated to get $\tau$, the full optical depth along a path.

In Earth convention, $d\tau = + n\sigma dz$ as we are looking up

In Astro convention, $d\tau = - n\sigma dz$ as we are looking down

$$ dI_\lambda = \frac{dE}{dA d\Omega dt d\lambda} $$
Same units as Planck function ($B_\nu$, $B_\lambda$).
$$ dI_\nu d \nu = dI_\lambda d\lambda $$
$F_\nu = \int I_\nu \cos \theta d\Omega$

## Radiative Transfer
### Solving the Radiative Transfer Equation
$$ \mu_\theta \frac{dI_\nu}{d\tau} = B_\nu - I_\nu $$
$$ \frac{dI_\nu}{d\tau} e^{\frac{\tau}{\mu_\theta}} = \frac{1}{\mu_\theta} \left( B_\nu e^{\frac{\tau}{\mu_\theta}} - I_\nu e^{\frac{\tau}{\mu_\theta}} \right) $$
$$ \left[ I_\nu e^{\frac{\tau}{\mu_\theta}} \right]^{\tau_1}_{\tau_0} = \int_{\tau_0}^{\tau_1} B_\nu e^{\frac{\tau}{\mu_\theta}} d\tau $$
$$ I_{\nu,2} = I_{\nu,1} e^{-\frac{\tau_2 - \tau_1}{\mu_\theta}} + \int_{\tau_1}^{\tau_2} B_\nu e^{-\frac{\tau_2 - \tau}{\mu_\theta}} \frac{d\tau}{\mu_\theta} $$
#### What if There is No Thermal Emmission
$$ B_\nu = 0 $$
$ I_{\nu,2} = I_{\nu,1} e^{-\delta \tau} $, $\delta \tau = \tau_2 - \tau_1$. This is known as the Beer-Lambert law.

#### What if the Gas is Isothermal
$$ B_\nu = \text{constant} $$
$$ I_{\nu,2} = I_{\nu,1} e^{-\delta \tau} + B_\nu \left( 1 - e^{-\delta \tau} \right) $$
If $ \delta \tau \ll 1$ (optically thick), $I_{\nu,2} \approx B_\nu$
#### What if the Gas is Isothermal and Optically Thin
If $ \delta \tau \gg 1$ (optically thin), $I_{\nu,2} \approx I_{\nu,1}$
### Where Does the Atmosphere Become Opaque
Photosphere = vertical Location where the atmosphere becomes optically thick, defined where the radiation has been absorbed by a factor of e (optical depth $\tau = 1$)

In general, the atmosphere becomes opaque at different heights for different wavelengths, because the absorption depends on the specific properties of the gas at each wavelength.

$$ d \tau_\nu = - \kappa_\nu \rho dz $$
$$ \frac{d\tau_\nu}{dz} = - \kappa_\nu \rho $$
$$ \tau_\nu(z) = \int^{z}_{\infty} -\kappa_\nu \rho dz $$
$$ P_{\tau = 1} = \frac{g}{\kappa_\nu} $$
$$ \kappa_\nu = \Sigma_i \frac{\rho_i}{\rho} \kappa_{\nu,i} $$
$\kappa_\nu $ is the weighted sum of the individual opacities of the different gas species, weighted by their respective mass fractions.
$\sigma_\nu = \frac{\kappa_\nu}{\rho} $ is the cross-section per unit mass.

## Lagrangian Perspective
The Lagrangian derivative describes the rate of change of a quantity as it moves with the flow. It is given by:
$$ \delta C = \frac{\partial C}{\partial t} \delta t + \frac{\partial C}{\partial \mathbf{x}} \cdot \delta \mathbf{x} + \frac{\partial C}{\partial \mathbf{y}} \cdot \delta \mathbf{y} + \frac{\partial C}{\partial \mathbf{z}} \cdot \delta \mathbf{z} $$
$$ \delta C = \frac{\partial C}{\partial t} \delta t + \left(u_x \frac{\partial C}{\partial x} + u_y \frac{\partial C}{\partial y} + u_z \frac{\partial C}{\partial z} \right) \delta t $$
$$\frac{DC}{Dt} = \frac{\partial C}{\partial t} + \mathbf{u} \cdot \nabla C$$

where $\mathbf{u}$ is the velocity field.
### Mass Conservation
$$ \frac{\partial M}{\partial t} = \frac{\partial}{\partial t}(\rho \delta x \delta y \delta z) = \frac{\partial \rho}{\partial t} \delta x \delta y \delta z $$
- net mass flux in $x$ direction $ = - \frac{\partial}{\partial x} (\rho u_x) \delta x \delta y \delta z $
- net mass flux into parcel $ = - \nabla \cdot (\rho \mathbf{u}) \delta x \delta y \delta z $
$$ \frac{\partial \rho}{\partial t} + \nabla \cdot (\rho \mathbf{u}) = 0 $$
$$ \frac{D \rho}{Dt} + \rho \nabla \cdot \mathbf{u} = 0 $$
